Purpose: Collect input about a vehicle from user, validate and display in clean format


```python
# Vehicle Superclass
class Vehicle:
    """Represent a general vehicle with a vehicle type."""

    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type
```

```python
# Automobile Subclass
class Automobile(Vehicle):
    """Represent an automobile with its specifications."""

    def __init__(self, year, make, model, doors, roof):
        super().__init__("car")

        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof

    def display_info(self):
        """Display the automobile's information."""

        print("\nVehicle Information")
        print("-------------------")
        print(f"Vehicle type: {self.vehicle_type}")
        print(f"Year: {self.year}")
        print(f"Make: {self.make}")
        print(f"Model: {self.model}")
        print(f"Number of doors: {self.doors}")
        print(f"Type of roof: {self.roof}")
```

## User Input and Validation

The following function collects vehicle information from
the user.

Input validation ensures that...
- The year is a positive integer.
- The number of doors is either 2 or 4.
- The roof type is either "solid" or "sun roof".
- Make and model are not empty.

```python
def main():
    """Collect automobile information and display the result."""

    print("Enter Vehicle Information")

    while True:
        try:
            year = int(input("Year: "))
            if year > 0:
                break
        except ValueError:
            pass
        print("Please enter a valid positive year.")

    while True:
        make = input("Make: ").strip()
        if make:
            break
        print("Make cannot be empty.")

    while True:
        model = input("Model: ").strip()
        if model:
            break
        print("Model cannot be empty.")

    while True:
        doors = input("Number of doors (2 or 4): ").strip()
        if doors in ("2", "4"):
            doors = int(doors)
            break
        print("Please enter 2 or 4.")

    while True:
        roof = input("Roof type (solid or sun roof): ").strip().lower()
        if roof in ("solid", "sun roof"):
            break
        print("Please enter 'solid' or 'sun roof'.")

    car = Automobile(year, make, model, doors, roof)
    car.display_info()
```

## Run the Application

Execute the following cell to enter information about a vehicle.

The program will then validate the input and display the vehicle's details.

```python
main()
```
